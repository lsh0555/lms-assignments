### chapter14 - 함수

함수(Function) : 관련된 코드를 하나의 이름으로 묶어 필요할 때 다시 사용할 수 있도록 만든 것

< 함수 장점 >

* 반복되는 코드를 줄일 수 있음
* 필요한 기능을 다시 사용 가능
* 코드의 역할을 구분 가능
* 긴 코드를 작은 기능으로 나눌 수 있음

< 함수 정의 >
`def` - define(정의하다)

def say_hello():
    print("Hello, Python!")


< 함수 호출 >
만들어진 함수를 실제로 실행하려면 함수 이름 뒤에 `()`를 붙인다.

say_hello()

def say_hello():  → 함수 정의
say_hello()       → 함수 호출

< 함수 이름 >
어떤 역할을 하는지 알 수 있도록 작성하는 것이 좋음 / `snake_case` 사용

def f():                   X
def show_student_info():   O

< 매개변수, 인수 >
함수를 실행할 때 필요한 값을 전달할 수 있다.

def greet(name):
    print(name)

greet("민수")

name     → 매개변수(parameter)
"민수"   → 인수(argument)

- 매개변수 : 정의할 때 값을 받을 자리 / 값을 받을 자리

def greet(name):


- 인수 : 호출할 때 실제로 전달하는 값 / 실제로 전달하는 값

greet("민수")


+ 여러 개의 매개변수
함수는 여러 값을 받을 수 있음

def show_profile(name, age, city):
    print(name)
    print(age)
    print(city)

show_profile("민수", 20, "서울")


+ return : 함수에서 처리한 결과를 함수 밖으로 돌려주는 역할
    `return`으로 받은 값은 변수에 저장하거나 다른 계산에서 다시 사용 가능

def add(a, b):
    return a + b

result = add(10, 20)


+ 여러 값 return
함수는 여러 개의 값을 반환 가능

def calculate(a, b):
    total = a + b
    difference = a - b

    return total, difference

이를 튜플 언패킹과 연결해서 가능

< 기본값 매개변수 >
매개변수에 기본값을 설정 가능 / 인수를 전달하지 않았을 때 기본값이 사용됨

def greet(name, message="안녕하세요"):
    print(message, name)

1) greet("민수")                      // message 전달 X
 -> 기본값인 `"안녕하세요"`가 사용된다.

2) greet("민수", "반갑습니다")         // message 전달
 -> 전달한 `"반갑습니다"`가 사용된다.


- 키워드 인수 : 함수를 호출할 때 매개변수의 이름을 직접 지정해서 값을 전달 가능
            어떤 값이 어떤 매개변수에 들어가는지 명확하게 알 수 있음

show_profile(
    name="민수",
    age=20,
    city="서울"
)


- 지역 변수(Local Variable) : 함수 안에서 만들어진 변수
      함수 내부에서 만든 지역 변수는 기본적으로 함수 내부에서 사용
      함수 밖에서도 결과가 필요하다면 `return` 사용

def calculate_total(price, quantity):
    total = price * quantity
    return total

price, quantity → 함수가 받은 값
total           → 함수 안의 지역 변수


result = calculate_total(12000, 3)    -> `result`는 함수 밖 변수

< 함수 안 조건문 >
함수 안에서도 `if`, `elif`, `else` 사용 가능
새로운 조건문 규칙 생기지 않으며, 조건문을 하나의 기능으로 묶은 것일 뿐

def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "C"


< 함수 안 반복문 - for문 >

def calculate_total(scores):
    total = 0

    for score in scores:
        total += score

    return total

+ 함수 안에서 리스트, 반복문, 조건문, 딕셔너리 등 사용 가능


###  긴 코드를 함수로 나누기

+ 모든 코드를 무조건 함수로 만들 필요는 없다

### chapter15 - Python 모듈·패키지와 가상환경

함수 → 하나의 역할을 코드로 묶음
모듈 → 관련 함수와 변수를 하나의 .py 파일로 묶음
패키지 → 관련 모듈을 폴더로 묶음

## 1. 모듈(Module) - 파이썬 코드를 담고 있는 하나의 `.py` 파일
 - 프로그램이 커지면 모든 코드를 하나의 파일에 작성하기보다 관련된 함수와 변수를 별도의 파일로 나누어 관리할 수 있다.

 - 직접 만든 `.py` 파일도 다른 파일에서 불러와 사용할 수 있는 모듈이 된다.


## 2. import : 다른 모듈의 기능을 현재 코드에서 사용할 수 있도록 불러오는 문법

### 모듈 전체 불러오기

import math
print(math.sqrt(16))

`math.sqrt()`는 `math` 모듈의 `sqrt` 기능을 사용한다는 의미이다.


### 별칭 사용하기

import statistics as stats   -> stats라는 별칭
print(stats.mean([80, 90, 100]))


## 3. 패키지(Package) : 관련된 여러 모듈을 폴더 구조로 묶어 관리하는 것**이다.

shop/
├── __init__.py
├── price.py
└── delivery.py

- `price.py` → 모듈
- `delivery.py` → 모듈
- `shop/` → 패키지


## 4. 표준 라이브러리 - Python과 함께 기본적으로 제공되는 기능

대표적인 모듈:
math
statistics
random
datetime
 ㄴ 별도의 `pip install` 없이 `import`하여 사용 가능

## 5. 외부 패키지 : 별도로 설치해서 사용하는 패키지

`requests` 패키지 설치 ->   python -m pip install requests

## 6. pip와 import 차이

`pip`와 `import`는 역할 다름

pip → 패키지를 설치
import → 설치된 기능을 코드에서 불러오기

 + python -m pip -> 현재 선택한 Python과 연결된 pip를 사용한다는 의미 명확해짐

### pip 버전 확인

python -m pip --version

### 설치된 패키지 목록 확인

python -m pip list

### 특정 패키지 확인

python -m pip show requests

## 8. 가상환경(Virtual Environment) : 프로젝트마다 Python 인터프리터와 패키지 설치 공간을 분리하는 방법

 * 프로젝트마다 필요한 패키지와 버전이 다를 수 있기 때문에 가상환경을 사용하면 충돌을 줄일 수 있다.

```text
프로젝트 A
└── .venv
    └── package X 버전 1

프로젝트 B
└── .venv
    └── package X 버전 2
```

## 9. .venv 생성

### Windows

```bash
py -m venv .venv
```

### macOS / Linux

```bash
python3 -m venv .venv
```

생성 후:

```text
my_project/
├── .venv/
└── main.py
```

`.venv`는 Python 코드 파일 X, 프로젝트 전용 실행 환경을 담고 있는 폴더


## 10. 가상환경 활성화

### Windows PowerShell

```powershell
.venv\Scripts\Activate.ps1
```

### Windows CMD

```text
.venv\Scripts\activate.bat
```

### macOS / Linux

```bash
source .venv/bin/activate
```

활성화되면 보통 터미널 앞에 다음과 같이 표시된다.

```text
(.venv)
```

가상환경 종료:

```bash
deactivate
```

## 11. VS Code Interpreter

가상환경을 만들었다고 해서 VS Code가 항상 자동으로 해당 환경을 사용하는 것은 아니다.

VS Code에서 사용할 Python 환경을 직접 선택할 수 있다.

```text
Ctrl + Shift + P
↓
Python: Select Interpreter
↓
프로젝트의 .venv 선택
```

### 중요

> 패키지를 설치한 Python과 코드를 실행하는 Python이 같은 환경인지 확인해야 한다.

## 13. .venv와 Git

`.venv`에는 설치된 패키지 등 많은 파일이 들어 있고 다시 만들 수 있기 때문에 일반적으로 Git에 올리지 않는다.

git에 올리지 않기 위해 `.gitignore`에 이 부분 추가  -> .venv/

대신 프로젝트에 필요한 패키지 정보는 `requirements.txt`에 기록한다.

## 14. requirements.txt : 프로젝트에 필요한 패키지와 버전을 기록하는 파일

### 현재 설치된 패키지 확인

python -m pip freeze


### requirements.txt 생성

python -m pip freeze > requirements.txt

### requirements.txt를 이용하여 패키지 설치

python -m pip install -r requirements.txt

### 차이

```text
.venv
→ 실제 패키지가 설치되어 있는 실행 환경

requirements.txt
→ 실행 환경을 다시 만들기 위한 패키지 목록
```

## 15. ModuleNotFoundError : Python이 필요한 모듈을 찾지 못했을 때 발생

오류가 발생했다고 바로 다시 설치하지 말고 다음 순서로 확인한다.

1. `import`한 이름이 정확한가?
2. 직접 만든 모듈이라면 파일 위치가 정확한가?
3. 외부 패키지가 현재 환경에 설치되어 있는가?
4. 현재 실행 중인 Python의 경로가 어디인가?
5. VS Code에서 같은 Interpreter를 선택했는가?
