## 핵심 복습 포인트


```python
# 1. 문자열 데이터형
value = "20"
print(type(value))  # str

# 2. 딕셔너리 값 조회
student = {"name": "민수", "score": 85}
print(student["score"])

# 3. 빈 집합
empty_set = set()
print(type(empty_set))

# 4. 빈 딕셔너리
empty_dict = {}
print(type(empty_dict))
```

```python
# 5. ValueError 예시
try:
    int("abc")
except ValueError as e:
    print(type(e).__name__)
```

```python
# 6. 함수의 매개변수
def add(a, b):
    return a + b

# a, b가 매개변수(parameter)
print(add(3, 5))
```

```python
# 7. 파일 모드
# r : 읽기
# w : 쓰기
# a : 기존 내용 뒤에 추가
```

```python
# 8. 클래스의 초기화 메서드
class Student:
    def __init__(self, name):
        self.name = name

student = Student("민수")
print(student.name)
```

## 암기 포인트

- `str` → 문자열 데이터형
- `student["score"]` → 딕셔너리 값 접근
- `set()` → 빈 집합
- `{}` → 빈 딕셔너리
- `ValueError` → 값을 원하는 자료형으로 변환할 수 없을 때 발생 가능
- `parameter` → 함수 정의에서 값을 받아들이는 이름
- `requirements.txt` → 패키지 목록 기록 파일
- `a` → 파일 뒤에 내용 추가
- `__init__()` → 객체 생성 시 초기값 설정
- `IPO` → Input / Process / Output
- 큰 기능을 역할별로 나눌 때 대표적으로 **함수**를 사용
